```python
#Q1 — Basic Dictionary Creation
#Create a dictionary called student with keys: name, age, course, city. Print each value.
#student = {
#"name": "",
#"age": ,
#"course": "",
#"city": "" }

student = {
    "name": "Abdullah",
    "age": 20,
    "course": "Artificial Intelligence",
    "city": "Karachi"
}

print("Name:", student["name"])
print("Age:", student["age"])
print("Course:", student["course"])
print("City:", student["city"])
```

```python

#Q2 — Adding & Updating Values
#car = {
#"brand": "Toyota",
#"model": "Corolla",
#"year": 2018 }
# Add color='White', update year to 2022 and print dictionary.

car = {
    "brand": "Toyota",
    "model": "Corolla",
    "year": 2018
}

# Add color
car["color"] = "White"

# Update year
car["year"] = 2022

print(car)
```

```python
#Q3 — Loop Through Dictionary
#person = {
#"name": "Ali",
#"age": 25,
#"city": "Karachi" }
#Print keys and values using loop

person = {
    "name": "Ali",
    "age": 25,
    "city": "Karachi"
}

# Print keys and values
for key, value in person.items():
    print(key, ":", value)
```

```python
#Q4 — Count Frequency of Numbers
#numbers = [1,2,2,3,3,3,4,4,4,4]
#Create frequency dictionary.

numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

frequency = {}

# Count each number
for number in numbers:
    if number in frequency:
        frequency[number] += 1
    else:
        frequency[number] = 1

print(frequency)
```

```python
#Q5 — Word Counter.Take sentence input and count word frequency.

sentence = input("Enter a sentence: ")

words = sentence.split()
word_count = {}

# Count each word
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

print(word_count)
```

```python
#Q6 — Student Marks & Average marks = {"Math":80, "English":70, "Physics":90} Calculate total and average.

marks = {
    "Math": 80,
    "English": 70,
    "Physics": 90
}

# Calculate total
total = sum(marks.values())

# Calculate average
average = total / len(marks)

print("Total:", total)
print("Average:", average)
```

```python
#Q7 — Highest Scoring Student
students = {
"Ali":85,
"Sara":92,
"Ahmed":78,
"Zara":95 }


highest_student = ""
highest_marks = 0

for name, marks in students.items():
    if marks > highest_marks:
        highest_marks = marks
        highest_student = name

print("Highest Scoring Student:", highest_student)
print("Marks:", highest_marks)
```

```python
# Q8 — Merge Dictionaries. Merge and print result.

dict1 = {"a":1,"b":2}
dict2 = {"c":3,"d":4}

merged_dict = dict1.copy()
merged_dict.update(dict2)

print(merged_dict)
```
