```python
# 1. Write a program that accepts a list from user and print the alternate element of list

numbers = list(map(int, input("Enter list elements separated by space: ").split()))

print("Alternate elements:")

for i in range(0, len(numbers), 2):
    print(numbers[i], end=" ")

```

```python

#2. Write a program that accepts a list from user. Your program should reverse the content of list and display it. Do not use reverse() method. 

numbers = list(map(int, input("\n\nEnter list elements separated by space: ").split()))

reversed_list = []

for i in range(len(numbers) - 1, -1, -1):
    reversed_list.append(numbers[i]) 

print("Reversed list:", reversed_list)

```

```python

#  3. Find and display the largest number of a list without using built-in function max(). Your program should ask the user to input values in list from keyboard

numbers = list(map(int, input("\nEnter list elements separated by space: ").split()))

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print("Largest number:", largest)


```

```python
#  4. Write a program that rotates the element of a list so that the element at the first index moves to the second index, the element in the second index moves to the third index, etc., and the element in the last index moves to the first index

numbers = list(map(int, input("\nEnter list elements separated by space: ").split()))

if len(numbers) > 0:
    rotated_list = [numbers[-1]]

    for i in range(len(numbers) - 1):
        rotated_list.append(numbers[i])

    print("Rotated list:", rotated_list)
else: 
    print("List is empty.")
```

```python

# 5. Write a program that input a string and ask user to delete a given word from a string

text = input("\nEnter a string: ")
word = input("Enter the word to delete: ")

words = text.split()
new_words = []

for w in words:
    if w != word:
        new_words.append(w)

new_string = " ".join(new_words)

print("String after deleting word:", new_string)
```

```python
# 6. Write a program that reads a string from the user containing a date in the form mm/dd/yyyy. It should print the date in the form March 12, 2021. 


date = input("\nEnter date (mm/dd/yyyy): ")

month, day, year = date.split("/")

months = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

month_name = months[int(month) - 1]

print("Formatted date:", month_name, int(day), ",", year)
```

```python
#  7. Write a program with a function that accepts a string from keyboard and create a new string after converting character of each word capitalized. For instance, if the sentence is "stop and smell the roses."the output should be "Stop And Smell The Roses"

def capitalize_words(sentence):
    words = sentence.split()
    new_words = []

    for word in words:
        new_words.append(word.capitalize())

    return " ".join(new_words)


sentence = input("\nEnter a sentence: ")

result = capitalize_words(sentence)

print("Capitalized sentence:", result)
```

```python

# 8. Find the sum of each row of matrix of size m x n. For example for the following matrix output will be like this : 
# 2 11 7 12
# 5 2  9 15
# 8 3 10 42
# Sum of row 1 = 32
# Sum of row 2 = 31
# Sum of row 3 = 63

rows = int(input("\nEnter number of rows: "))
columns = int(input("Enter number of columns: "))

matrix = []

print("Enter matrix elements:")

for i in range(rows):
    row = list(map(int, input(f"Enter row {i + 1}: ").split()))
    matrix.append(row)

for i in range(rows):
    row_sum = 0

    for j in range(columns):
        row_sum += matrix[i][j]

    print(f"Sum of row {i + 1} =", row_sum)
```

```python

# 9. Write a program to add two matrices of size n x m. 

rows = int(input("\nEnter number of rows: "))
columns = int(input("Enter number of columns: "))

print("Enter first matrix:")

matrix1 = []

for i in range(rows):
    row = list(map(int, input(f"Enter row {i + 1}: ").split()))
    matrix1.append(row)

print("Enter second matrix:")

matrix2 = []

for i in range(rows):
    row = list(map(int, input(f"Enter row {i + 1}: ").split()))
    matrix2.append(row)

result = []

for i in range(rows):
    row = []

    for j in range(columns):
        row.append(matrix1[i][j] + matrix2[i][j])

    result.append(row)

print("Sum of two matrices:")

for row in result:
    print(row)
```

```python

# 10: Write a program to multiply two matrices.

rows1 = int(input("\nEnter rows of first matrix: "))
columns1 = int(input("Enter columns of first matrix: "))

print("Enter first matrix:")

matrix1 = []

for i in range(rows1):
    row = list(map(int, input(f"Enter row {i + 1}: ").split()))
    matrix1.append(row)


rows2 = int(input("Enter rows of second matrix: "))
columns2 = int(input("Enter columns of second matrix: "))

print("Enter second matrix:")

matrix2 = []

for i in range(rows2):
    row = list(map(int, input(f"Enter row {i + 1}: ").split()))
    matrix2.append(row)


# Matrix multiplication is possible only when
# columns of first matrix = rows of second matrix

if columns1 != rows2:
    print("Matrix multiplication is not possible.")

else:
    result = []

    for i in range(rows1):
        row = []

        for j in range(columns2):
            total = 0

            for k in range(columns1):
                total += matrix1[i][k] * matrix2[k][j]

            row.append(total)

        result.append(row)

    print("Multiplication of two matrices:")

    for row in result:
        print(row)
```
